```python
"""
================================================================
Regressione Lineare Multipla con Scikit-learn
================================================================
Esempio didattico: creazione, addestramento e valutazione
di un modello di regressione lineare multipla.
"""

# --- STEP 1: Importazione delle librerie necessarie ---
# Scikit-learn per il modello e le metriche di valutazione,
# NumPy e Pandas per la gestione dei dati,
# Matplotlib per la visualizzazione grafica dei risultati.
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# --- STEP 2: Creazione del dataset di esempio ---
# Due feature numeriche correlate linearmente con la variabile target.
X = pd.DataFrame({
    'Feature1': [1, 2, 3, 4, 5, 6, 7, 8],
    'Feature2': [2, 4, 6, 8, 10, 12, 14, 16]
})

y = np.array([1.1, 2.3, 3.0, 3.8, 5.2, 6.1, 6.9, 8.3])


# --- STEP 3: Creazione e addestramento del modello ---
# Il metodo fit() calcola i coefficienti ottimali che minimizzano
# la somma dei quadrati degli errori tra valori reali e predetti.
model = LinearRegression()
model.fit(X, y)


# --- STEP 4: Analisi di intercetta e coefficienti ---
# intercept_ : valore della retta quando tutte le feature sono zero.
# coef_      : peso associato a ciascuna variabile indipendente.
print("Intercetta (intercept):", model.intercept_)
print("Coefficienti (coefficients):", model.coef_)


# --- STEP 5: Predizioni e valutazione del modello ---
# predict() genera i valori stimati dal modello.
# R² misura la percentuale di varianza spiegata (piu' vicino a 1, meglio e').
# MSE quantifica l'entità media degli errori (piu' basso, meglio e').
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print("Valori predetti:", y_pred)
print("Coefficiente R²:", round(r2, 4))
print("Errore quadratico medio (MSE):", round(mse, 4))


# --- STEP 6: Visualizzazione dei risultati ---
# Un grafico a dispersione confronta visivamente i valori reali
# con quelli predetti dal modello. Il grafico viene esportato
# anche come immagine PNG, utile per la documentazione.
plt.figure(figsize=(8, 5))
plt.scatter(range(len(y)), y, color='blue', label='Valori reali')
plt.plot(range(len(y_pred)), y_pred, color='red', marker='o', label='Valori predetti')
plt.title('Regressione Lineare: Valori Reali vs Predetti')
plt.xlabel('Indice campione')
plt.ylabel('Valore target')
plt.legend()
plt.grid(True)
plt.savefig('regressione_lineare_sklearn.png', dpi=150, bbox_inches='tight')
plt.show()
```

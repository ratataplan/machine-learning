```python
"""
Regressione Lineare Semplice con NumPy
----------------------------------------
Questo script implementa da zero (senza scikit-learn) una regressione
lineare semplice tramite il metodo dei minimi quadrati (Ordinary Least
Squares), per mostrare la matematica che si nasconde dietro al modello.
"""

import numpy as np
import matplotlib.pyplot as plt

# STEP 1: Dataset di esempio
# X rappresenta la variabile indipendente (feature)
# y rappresenta la variabile dipendente (target) da predire
X = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([1.2, 2.3, 2.9, 4.1, 5.3, 5.8, 7.2, 7.9])

# STEP 2: Calcolo delle medie di X e y
# Le medie sono necessarie per calcolare la retta di regressione
# tramite il metodo dei minimi quadrati (least squares)
X_mean = np.mean(X)
y_mean = np.mean(y)

# STEP 3: Calcolo dei coefficienti della retta (beta_1 e beta_0)
# beta_1 rappresenta la pendenza (slope) della retta di regressione
# beta_0 rappresenta l'intercetta (intercept) della retta di regressione
numerator = np.sum((X - X_mean) * (y - y_mean))
denominator = np.sum((X - X_mean) ** 2)

beta_1 = numerator / denominator
beta_0 = y_mean - beta_1 * X_mean

print(f"Coefficiente angolare (beta_1): {beta_1:.4f}")
print(f"Intercetta (beta_0): {beta_0:.4f}")

# STEP 4: Calcolo dei valori predetti dal modello
y_pred = beta_0 + beta_1 * X

# STEP 5: Valutazione del modello tramite R^2 (coefficiente di determinazione)
# R^2 misura quanto bene il modello spiega la varianza dei dati osservati.
# Un valore vicino a 1 indica un ottimo adattamento del modello ai dati.
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y_mean) ** 2)
r_squared = 1 - (ss_res / ss_tot)

print(f"R^2 (coefficiente di determinazione): {r_squared:.4f}")

# STEP 6: Visualizzazione grafica dei dati e della retta di regressione
plt.figure(figsize=(8, 6))
plt.scatter(X, y, label="Dati osservati", color="blue")
plt.plot(X, y_pred, label="Retta di regressione", color="red", linewidth=2)
plt.xlabel("X (variabile indipendente)")
plt.ylabel("y (variabile dipendente)")
plt.title("Regressione Lineare Semplice con NumPy")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("regressione_lineare.png", dpi=150)
plt.show()
```

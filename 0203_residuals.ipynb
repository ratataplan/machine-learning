```python
"""
Analisi dei Residui in un Modello di Regressione Lineare
-----------------------------------------------------------
Questo script mostra come calcolare e visualizzare i residui
(differenza tra valori osservati e valori previsti) di un modello
di regressione lineare addestrato con Scikit-learn, utilizzando
come esempio la previsione del prezzo di un'abitazione in base
alla sua superficie.
"""

# STEP 1 - Importazione delle librerie e creazione del dataset
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Dataset di esempio: superficie (m²) e prezzo (migliaia di euro) di 10 abitazioni
superficie = np.array([50, 60, 65, 70, 80, 85, 90, 100, 110, 120]).reshape(-1, 1)
prezzo = np.array([120, 135, 140, 150, 165, 172, 180, 195, 210, 225])


# STEP 2 - Creazione e addestramento del modello di regressione lineare
modello = LinearRegression()
modello.fit(superficie, prezzo)

print(f"Coefficiente angolare (pendenza): {modello.coef_[0]:.2f}")
print(f"Intercetta: {modello.intercept_:.2f}")


# STEP 3 - Calcolo delle previsioni del modello sui dati di addestramento
prezzo_previsto = modello.predict(superficie)


# STEP 4 - Calcolo dei residui (valori osservati meno valori previsti)
residui = prezzo - prezzo_previsto

print("\nResidui calcolati per ciascuna abitazione:")
for i, r in enumerate(residui):
    print(f"  Abitazione {i + 1}: residuo = {r:.2f} (migliaia di euro)")


# STEP 5 - Creazione del grafico dei residui
plt.figure(figsize=(8, 6))
plt.scatter(prezzo_previsto, residui, color="steelblue", edgecolor="black", s=70)
plt.axhline(0, color="red", linestyle="--", linewidth=1.5)
plt.xlabel("Valori Previsti (migliaia di euro)")
plt.ylabel("Residui (migliaia di euro)")
plt.title("Grafico dei Residui - Modello Prezzo Abitazioni")
plt.grid(alpha=0.3)

# Esportazione del grafico in formato PNG
plt.savefig("grafico_residui.png", dpi=150, bbox_inches="tight")
plt.show()


# STEP 6 - Analisi statistica riassuntiva dei residui
media_residui = np.mean(residui)
deviazione_residui = np.std(residui)

print(f"\nMedia dei residui: {media_residui:.2f}")
print(f"Deviazione standard dei residui: {deviazione_residui:.2f}")

if abs(media_residui) < 1e-6:
    print("La media dei residui è pari a zero: comportamento atteso per una regressione OLS.")
```
